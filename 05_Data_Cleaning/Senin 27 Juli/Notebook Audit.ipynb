Notebook ini melakukan audit awal terhadap dataset `movies_metadata.csv` (The Movies Dataset dari Kaggle) sebelum masuk ke tahap cleaning. Audit mencakup: jumlah data, jumlah kolom, tipe data, missing value, dan duplicate.

> **Catatan:** jalankan notebook ini di folder yang sama dengan file `movies_metadata.csv` hasil download dari Kaggle. Angka pembanding di setiap sel markdown diambil dari hasil eksplorasi Minggu 2 (Dataset Summary), dipakai sebagai referensi untuk verifikasi ulang hasil audit hari ini.

```python
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv("movies_metadata.csv", low_memory=False)
print("Dataset berhasil dimuat.")
```

## 1. Jumlah Data & Kolom

```python
print(f"Jumlah baris  : {df.shape[0]:,}")
print(f"Jumlah kolom  : {df.shape[1]}")
```

**Referensi Minggu 2:** 45.466 baris, 24 kolom.

## 2. Info Umum & Tipe Data

```python
df.info()
```

## 3. Missing Value per Kolom

```python
missing = df.isnull().sum()
missing_pct = (missing / len(df) * 100).round(2)

missing_summary = pd.DataFrame({
    "Jumlah_Missing": missing,
    "Persentase(%)": missing_pct
})
missing_summary = missing_summary[missing_summary["Jumlah_Missing"] > 0].sort_values(
    "Jumlah_Missing", ascending=False
)
missing_summary
```

**Referensi Minggu 2:** 20 dari 24 kolom punya missing value. Terbanyak: `belongs_to_collection` (90,12%), `homepage` (82,88%), `tagline` (55,10%).

## 4. Duplicate Data

```python
jumlah_duplikat_penuh = df.duplicated().sum()
jumlah_id_duplikat = df["id"].duplicated().sum()
jumlah_judul_duplikat = df["title"].duplicated().sum()

print(f"Baris duplikat penuh  : {jumlah_duplikat_penuh}")
print(f"ID duplikat           : {jumlah_id_duplikat}")
print(f"Judul (title) duplikat: {jumlah_judul_duplikat}")
```

**Referensi Minggu 2:** 17 baris duplikat penuh, 30 ID duplikat, 3.188 judul duplikat.

## 5. Statistik Deskriptif (Kolom Numerik)

```python
df.describe()
```

**Catatan penting:** kolom `budget` dan `revenue` memiliki Q1=Q3=0 karena mayoritas nilainya 0 — ini indikasi data tidak dilaporkan, bukan variasi alami. Perlu ditangani sebagai missing value (bukan angka 0 asli) di tahap cleaning Selasa.

## 6. Ringkasan Permasalahan yang Ditemukan

1. **Missing Value** — 20 dari 24 kolom, terbanyak `belongs_to_collection`, `homepage`, `tagline`.
2. **Duplicate Data** — 17 baris duplikat penuh, 3.188 judul duplikat.
3. **Format JSON String** — kolom `genres`, `production_companies`, `production_countries`, `spoken_languages` masih berbentuk kode JSON, perlu diparsing.
4. **Data Tidak Valid (Misalignment)** — kolom `adult` seharusnya cuma True/False, tapi ada baris berisi teks sinopsis (indikasi pergeseran kolom).
5. **Nilai 0 yang Menyesatkan** — `budget` dan `revenue` mayoritas bernilai 0, bukan Rp0 asli, melainkan data tidak dilaporkan.
6. **Outlier Awal** — budget, revenue, runtime, popularity, vote_average, vote_count semuanya punya outlier signifikan berdasarkan metode IQR (dicek lebih lanjut di hari-hari berikutnya).
