Menganalisis karakteristik dataset Titanic menggunakan statistik deskriptif.

```python
import pandas as pd

df = pd.read_csv('Final_Clean_Dataset_titanic.csv')
df.head()
```

## 1. describe() - Statistik Deskriptif Kolom Numerik

```python
df[['Age', 'Fare', 'Family_Size', 'Sib_Sp', 'Parch']].describe()
```

## 2. Statistik Dasar Satu per Satu (Mean, Median, Modus, Min, Max, Std, Kuartil)

```python
numeric_cols = ['Age', 'Fare', 'Family_Size', 'Sib_Sp', 'Parch']

for col in numeric_cols:
    s = df[col]
    print(f"--- {col} ---")
    print(f"Mean       : {s.mean():.2f}")
    print(f"Median     : {s.median():.2f}")
    print(f"Modus      : {s.mode().iloc[0]:.2f}")
    print(f"Minimum    : {s.min():.2f}")
    print(f"Maksimum   : {s.max():.2f}")
    print(f"Std Deviasi: {s.std():.2f}")
    print(f"Q1 (25%)   : {s.quantile(0.25):.2f}")
    print(f"Q3 (75%)   : {s.quantile(0.75):.2f}")
    print()
```

## 3. value_counts() - Distribusi Kolom Kategorikal

```python
print(df['Sex'].value_counts())
print()
print(df['Embarked'].value_counts())
print()
print(df['P_Class'].value_counts())
```

```python
print(df['Age_Group'].value_counts())
print()
print(df['Title'].value_counts())
print()
print(df['Survived'].value_counts())
```

## 4. unique() dan nunique() - Nilai Unik per Kolom

```python
categorical_cols = ['Sex', 'Embarked', 'P_Class', 'Age_Group', 'Title', 'Survived']

for col in categorical_cols:
    print(f"{col}: {df[col].nunique()} nilai unik -> {df[col].unique().tolist()}")
```
